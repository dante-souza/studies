# Mutabilidade

Por padrão, modelos Pydantic são mutáveis: podemos alterar os valores de uma instância depois de criada.

```python
from pydantic import BaseModel, ConfigDict, ValidationError
```

```python
class Model(BaseModel):
    field: int
```

```python
m = Model(field=10)
```

E podemos alterar a instância:

```python
m.field=20
```

```python
print(f"m: {m!r}")
```

Podemos controlar se o modelo é mutável ou não com `model_config`:

```python
class Model(BaseModel):
    model_config = ConfigDict(frozen=True)

    field: int
```

```python
m = Model(field=10)
```

```python
try:
    m.field=20
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

Além da implicação óbvia (modelos que sabemos ser imutáveis), um efeito colateral importante é que agora podemos usar a instância como
**chave de dicionário** (lembre que chaves de dicionário não podem ser objetos mutáveis/não-hashable).

```python
class Model(BaseModel):
    model_config = ConfigDict(frozen=False)  # o padrão

    field: int
```

```python
m = Model(field=10)
```

E se tentarmos usar `m` como chave de dicionário:

```python
try:
    d = {m: "not gonna work!"}
except TypeError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

Repare que não é erro de validação — a exceção não tem nada a ver com o Pydantic, e sim com os dicionários Python.

Se tornarmos o modelo imutável:

```python
class Model(BaseModel):
    model_config = ConfigDict(frozen=True)

    field: int
```

```python
m = Model(field=10)
```

Agora podemos usar `m` como chave de dicionário:

```python
d = {m: "all's good!"}
```

```python
print(f"d: {d!r}")
```
