# Combinando anulável e opcional

Como um campo pode ser:
- anulável ou não anulável
- obrigatório ou opcional

temos quatro combinações possíveis.

Pode parecer óbvio, mas às vezes confunde; então vamos ver cada combinação separadamente e entender o que significa em termos de validação.

| | não anulável | anulável |
|---|---|---|
| **obrigatório** | `field: int` | `field: int \| None` |
| **opcional** | `field: int = 0` | `field: int \| None = None` |

## Obrigatório, não anulável

É basicamente a forma padrão de definir campos:

```python
from pydantic import BaseModel, ValidationError
```

```python
class Model(BaseModel):
    field: int
```

Aqui `field` é obrigatório e só aceita inteiros (portanto, não anulável).

```python
try:
    Model()
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

```python
try:
    Model(field=None)
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

## Obrigatório, anulável

Para o campo ser obrigatório, simplesmente não damos default; e usamos o *type hint* para indicar que `None` é aceitável.

```python
class Model(BaseModel):
    field: int | None
```

```python
try:
    Model()
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

```python
print(f"Model(field=None): {Model(field=None)!r}")
```

## Opcional, não anulável

```python
class Model(BaseModel):
    field: int = 0
```

Como comentei numa aula anterior, cuidado para não usar `None` como default aqui, já que o *type hint* não diz que `None` é aceitável.
Por padrão o Pydantic deixa passar (porque não valida defaults), mas veremos numa próxima aula como mudar isso.

```python
print(f"Model(): {Model()!r}")
```

```python
try:
    Model(field=None)
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

## Opcional, anulável

E, por fim, a combinação mais comum para campos anuláveis: também opcionais, com `None` como default:

```python
class Model(BaseModel):
    field: int | None = None
```

```python
print(f"Model(): {Model()!r}")
```

```python
print(f"Model(field=None): {Model(field=None)!r}")
```
