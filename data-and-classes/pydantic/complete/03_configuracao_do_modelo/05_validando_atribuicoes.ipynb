# Validando atribuições

Como vimos, o Pydantic valida os dados ao desserializar, mas, por padrão, **não** valida quando alteramos os dados de uma instância existente:

```python
from pydantic import BaseModel, ConfigDict, ValidationError
```

```python
class Model(BaseModel):
    field: int
```

```python
m = Model(field=10)
```

```python
m.field = "Python"
```

```python
print(f"m: {m!r}")
```

Conseguimos mudar `field` para um valor que não respeita a especificação do campo.

Assim como não valida defaults, o Pydantic não valida atribuições depois que a instância existe. Na maioria dos casos tudo bem, porque
controlamos os valores que atribuímos. Mas às vezes queremos habilitar esse recurso.

Podemos habilitá-lo no nível do modelo (vale para todos os campos) com `model_config`:

```python
class Model(BaseModel):
    model_config = ConfigDict(validate_assignment=True)

    field: int
```

```python
m = Model(field=10)
```

```python
try:
    m.field = "Python"
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

> 💡 **Complemento:** com `validate_assignment`, a atribuição também passa pela **coerção** — `m.field = "20"` guarda o inteiro `20`.

```python
m.field = "20"
print(f"m.field: {m.field!r}")
```
