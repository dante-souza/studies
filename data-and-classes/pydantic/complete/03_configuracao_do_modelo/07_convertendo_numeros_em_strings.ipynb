# Convertendo números em strings

Já discutimos que objetos, embora tenham representação em string (`str()`), não são convertidos para string em modelos Pydantic.

```python
from pydantic import BaseModel, ConfigDict, ValidationError
```

```python
class Model(BaseModel):
    field: str
```

```python
try:
    Model(field=100)
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

Às vezes, porém, **queremos** que números sejam convertidos em strings.

Para isso, configuramos o modelo para habilitar a conversão automática de números para strings. Para outros tipos de objeto, podemos usar
validadores customizados, que veremos mais adiante.

```python
class Model(BaseModel):
    model_config = ConfigDict(coerce_numbers_to_str=True)

    field: str
```

E o exemplo anterior agora funciona:

```python
print(f"Model(field=100): {Model(field=100)!r}")
```
