# Geração de JSON Schema

Para documentar APIs REST, uma abordagem comum é usar Swagger (OpenAPI) para gerar a documentação automaticamente.

Para isso, ele recebe um JSON Schema da requisição (se houver) e da resposta (se houver) de cada endpoint (além de outros dados).

Escrever um JSON Schema à mão é difícil, mas felizmente o Pydantic faz isso por nós.

Se você desenvolve com FastAPI, nem vai precisar do que mostro aqui: o FastAPI se integra ao Pydantic e cuida dessa "encanação" automaticamente.

Ainda assim, se precisar gerar o JSON Schema manualmente (talvez um endpoint que devolve um schema solicitado), faça assim:

```python
from pydantic import BaseModel
```

```python
class Model(BaseModel):
    field_1: int | None = None
    field_2: str = "Python"
```

```python
from pprint import pprint

pprint(Model.model_json_schema())
```

Há algumas opções para ajustar a geração do JSON Schema, principalmente relacionadas a *aliases* de campo, que veremos depois.

A documentação está [aqui](https://docs.pydantic.dev/latest/api/base_model/#pydantic.main.BaseModel.model_json_schema).

Também há formas bem mais avançadas de trabalhar com JSON Schema, além do escopo do curso.
