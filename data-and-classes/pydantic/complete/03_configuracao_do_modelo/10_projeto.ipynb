Este é o modelo Pydantic com que terminamos o projeto da seção anterior:

```python
from datetime import date
from pydantic import BaseModel


class Automobile(BaseModel):
    manufacturer: str
    series_name: str
    type_: str
    is_electric: bool = False
    manufactured_date: date
    base_msrp_usd: float
    vin: str
    number_of_doors: int = 4
    registration_country: str | None = None
    license_plate: str | None = None
```

Modifique o modelo `Automobile` para:
- proibir campos extras
- remover espaços no início/fim de todos os campos string
- validar defaults e atribuições
- usar a enum abaixo no campo `type_`

```python
from enum import Enum

class AutomobileType(Enum):
    sedan = "Sedan"
    coupe = "Coupe"
    convertible = "Convertible"
    suv = "SUV"
    truck = "Truck"
```
