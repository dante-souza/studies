# Serialização

Vamos começar com o mesmo modelo de antes:

```python
from pydantic import BaseModel, ValidationError

class Person(BaseModel):
    first_name: str
    last_name: str
    age: int
```

E criar algumas instâncias:

```python
galois = Person(first_name='Evariste', last_name='Galois', age=20)
newton = Person(first_name='Isaac', last_name='Newton', age=84)
```

São objetos Python comuns — têm até dicionário de instância:

```python
print(f"newton.__dict__: {newton.__dict__!r}")
```

Mas, por herdarem de `BaseModel`, ganham muita funcionalidade extra.

Por exemplo, podemos gerar um dicionário ou uma string JSON que representa os dados da instância.

O Pydantic oferece dois métodos para isso:
- `model_dump()`
- `model_dump_json()`

Esses métodos pegam os dados da instância e produzem outro objeto — um `dict` ou uma `str`.

Isso se chama **serializar** o modelo.

```python
print(f"galois.model_dump(): {galois.model_dump()!r}")
```

Isso é um `dict` Python:

```python
print(f"type(galois.model_dump()): {type(galois.model_dump())!r}")
```

E podemos serializar para JSON:

```python
print(f"newton.model_dump_json(): {newton.model_dump_json()!r}")
```

Repare nas aspas em volta da saída acima: é uma string, não um dicionário. E a string contém JSON válido.

```python
print(f"type(newton.model_dump_json()): {type(newton.model_dump_json())!r}")
```

O `model_dump_json()` aceita alguns argumentos de formatação, como `indent`.

> 📝 **Nota de revisão:** o original dizia que, por baixo, o Pydantic usa `dumps()` do módulo `json`. Na v2 isso não é verdade:
> a serialização JSON é feita pelo `pydantic-core` (Rust). Os argumentos aceitos são os da própria assinatura de `model_dump_json()`
> (`indent`, `include`, `exclude`, `by_alias`, `exclude_none`...), não os de `json.dumps`.

```python
print(f"newton.model_dump_json(indent=2): {newton.model_dump_json(indent=2)!r}")
```

E podemos usar `print()` para que os caracteres de escape sejam exibidos corretamente:

```python
print(f"newton.model_dump_json(indent=2):\n{newton.model_dump_json(indent=2)}")
```

Também podemos escolher excluir campos da serialização com o argumento `exclude` dos métodos de dump:

```python
print(f"galois.model_dump(exclude=['first_name', 'age']): {galois.model_dump(exclude=['first_name', 'age'])!r}")
```

O mesmo vale para a versão JSON.

> 📝 **Nota de revisão:** no original, o exemplo "da versão JSON" chamava `model_dump` (dict). Abaixo, a versão JSON de fato:

```python
print(f"newton.model_dump_json(exclude=['age']): {newton.model_dump_json(exclude=['age'])!r}")
```

Há várias formas de controlar quais dados são serializados e até **como** são serializados. Veremos isso em aulas futuras.

Em vez de escolher campos para excluir, podemos escolher quais incluir (e todos os outros ficam de fora):

```python
print(f'newton.model_dump(include=["last_name"]): {newton.model_dump(include=["last_name"])!r}')
```
