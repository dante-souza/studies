# PositiveInt

Este tipo especializado restringe o campo a um inteiro **positivo** (estritamente maior que zero). Há variações como `NegativeInt`,
`NonNegativeInt` etc., e também as versões para float.

A lista completa está na documentação do Pydantic [aqui](https://docs.pydantic.dev/latest/api/types/).

```python
from pydantic import BaseModel, PositiveInt, ValidationError
```

```python
class Circle(BaseModel):
    center: tuple[int, int] = (0, 0)
    radius: PositiveInt = 1
```

Agora temos um modelo `Circle` que só aceita uma tupla de 2 inteiros como centro e um inteiro positivo como raio.

```python
print(f"Circle(): {Circle()!r}")
```

```python
print(f"Circle(center=(1, 1), radius=10): {Circle(center=(1, 1), radius=10)!r}")
```

```python
try:
    Circle(center=(0.5, 0.5), radius = 0)
except ValidationError as ex:
    print(f"{type(ex).__name__}: {ex}")
```

Vejamos como isso aparece nos campos do modelo:

```python
print(f"Circle.model_fields: {Circle.model_fields!r}")
```

Repare em como aparece a propriedade `radius`:
- há uma `annotation` (vamos aprofundar isso bastante depois)
- `metadata` com uma definição `Gt` (*greater than*, com valor 0 — ou seja, maior que 0)

Veremos depois que dá para definir esse `PositiveInt` diretamente com anotações e restrições do `Field`.

Por ora, guarde que `PositiveInt`, e muitos outros tipos do Pydantic (não todos), são na verdade **tipos anotados** (`Annotated`).

```python
from typing import get_args
print(f"get_args(PositiveInt): {get_args(PositiveInt)!r}")
```
