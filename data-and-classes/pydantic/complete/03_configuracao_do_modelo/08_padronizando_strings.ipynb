# Padronizando strings

Ao lidar com strings, muitas vezes precisamos nos preocupar com maiúsculas/minúsculas ou espaços no início/fim.

Por exemplo, se recebêssemos estas duas strings nos dados:

```python
s1 = "   python"
s2 = " python   \t"
```

Obviamente, elas não são iguais:

```python
print(f"s1 == s2: {s1 == s2!r}")
```

A pergunta é: queremos que `s1` e `s2` sejam consideradas iguais?

Normalmente, sim — queremos "ignorar" os espaços.

Em Python puro faríamos assim:

```python
print(f"s1.strip() == s2.strip(): {s1.strip() == s2.strip()!r}")
```

Poderíamos usar validadores customizados, mas o Pydantic permite configurar isso no nível do modelo:

```python
from pydantic import BaseModel, ConfigDict
```

```python
class Model(BaseModel):
    model_config = ConfigDict(str_strip_whitespace=True)

    field: str
```

```python
m1 = Model(field="   python")
m2 = Model(field="  python   \t")
```

```python
print(f"m1: {m1!r}")
```

```python
print(f"m2: {m2!r}")
```

```python
print(f"m1 == m2: {m1 == m2!r}")
```

Os espaços foram removidos e agora comparamos as strings facilmente.

Outra questão comum é padronizar certos campos string em minúsculas ou maiúsculas.

O Pydantic permite especificar que todas as strings do modelo sejam convertidas para minúsculas ou maiúsculas.

Para outra padronização, teremos de usar validadores customizados (veremos em breve).

Mas, dependendo da necessidade, a conversão automática embutida pode ajudar.

```python
class Model(BaseModel):
    model_config=ConfigDict(str_to_lower=True) # combinado com str_to_upper, o lower prevalece

    field: str
```

```python
m = Model(field="PYTHON")
print(f"m: {m!r}")
```

A string foi convertida para minúsculas na desserialização.

Também podemos escolher maiúsculas:

```python
class Model(BaseModel):
    model_config=ConfigDict(str_to_upper=True)

    field: str
```

```python
m = Model(field="Python")
print(f"m: {m!r}")
```

Conferindo a afirmação do comentário acima (com as duas opções ligadas, o lower prevalece):

```python
class Model(BaseModel):
    model_config = ConfigDict(str_to_lower=True, str_to_upper=True)

    field: str
```

```python
print(f'Model(field="PyThOn"): {Model(field="PyThOn")!r}')
```
